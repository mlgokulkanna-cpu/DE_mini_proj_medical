```python
%sql
CREATE SCHEMA IF NOT EXISTS medical_pipeline.bronze;
```

```python

src_catalog = "workspace"
bronze_catalog = "medical_pipeline"
source_schema = "azure_blob_storage"
bronze_schema = "bronze"



tables = ["data_dictionary", "patients", "encounters", "organizations", "payers", "procedures"]

for table in tables:
    df = spark.table(f"{src_catalog}.{source_schema}.{table}")
    
    df.write.format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"{bronze_catalog}.bronze.{table}")

print("Bronze Delta tables created")
```
