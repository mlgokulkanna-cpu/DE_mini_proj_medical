```python
%sql
CREATE SCHEMA IF NOT EXISTS medical_pipeline.silver;
```

```python
from pyspark.sql.functions import col
import re
from pyspark.sql.functions import lower, trim
df_silver_payers = spark.table("medical_pipeline.bronze.payers")
df_silver_payers = df_silver_payers.select("id", "name", "address", "city", "state_headquartered","zip")
df_silver_payers = df_silver_payers.withColumn("id", col("id").cast("string"))
df_silver_payers = df_silver_payers.dropna(how="any")
df_silver_payers = df_silver_payers.dropDuplicates(["id"])
df_silver_payers = df_silver_payers.withColumn("name", lower(trim(col("name"))))
df_silver_payers = df_silver_payers.filter(col("id").isNotNull())
     
```

```python
df_silver_payers.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("medical_pipeline.silver.payers")
```
