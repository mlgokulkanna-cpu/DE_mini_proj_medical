```python
%sql
create or replace table medical_pipeline.gold.kpi_encountrs_percent as 
with table1 as (
select  month(start) as month ,  sum(more_than_24) as more_than_24 ,count(*)-sum(more_than_24) as less_than_24 
 from medical_pipeline.gold.fact_encounters group by month(start))
 select * , (more_than_24/less_than_24)*100 as more_than_24_percent from table1;
```
