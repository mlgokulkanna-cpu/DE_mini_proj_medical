```python
%sql
select
  encounter_class,
  quarter(start) as quarter,
  count(*)
    / (
      select
        count(*)
      from
        medical_pipeline.gold.fact_encounters
    )*100 as percentage
from
  medical_pipeline.gold.fact_encounters   
group by
  encounter_class,
  quarter(start)
order by
  encounter_class,
  quarter(start)
```
