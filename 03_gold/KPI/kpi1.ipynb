```python
%sql
create or replace table medical_pipeline.gold.kpi_encounter_mix as 
select
  encounter_class,
  year(start) as year,
  quarter(start) as quarter,
  count(*)
    / (
      select
        count(*)
      from
        medical_pipeline.gold.fact_encounters
    )*100 as percentage
from
  medical_pipeline.gold.fact_encounters   
group by
  encounter_class,
  year(start),
  quarter(start)

order by
  encounter_class,
  year(start),
  quarter(start)
```

```python
%sql
select * from medical_pipeline.gold.kpi_encounter_mix
```
