```python
%sql
create or replace table medical_pipeline.gold.kpi_zero_coverage_percentage_monthly as
with table1 as(
select
payer_id,
  year(start) as year,
  month(start) as month,
  count(*) as total,
  sum(has_payer_coverage) as covered_by_payer,
  (sum(has_payer_coverage) / count(*)) * 100 as covered_percentage
from
  medical_pipeline.gold.fact_encounters
group by
  payer_id, year(start),month(start)
order by
  year(start),month(start)
),table2 as(
  select dp.payer_id,dp.payer_name,year,table1.month, total,covered_by_payer,covered_percentage from medical_pipeline.gold.dim_payers dp join table1 on dp.payer_id = table1.payer_id order by year,table1.month
)select * from table2
```
