```python
%sql

CREATE OR REPLACE TABLE medical_pipeline.gold.kpi_avg_claim_cost_by_payer AS
SELECT 
  p.payer_id,
  p.payer_name,
  AVG(f.total_cost) AS avg_claim_cost
FROM medical_pipeline.gold.fact_encounters f
LEFT JOIN medical_pipeline.gold.dim_payers p
  ON f.payer_id = p.payer_id
GROUP BY p.payer_id, p.payer_name
ORDER BY avg_claim_cost DESC;

SELECT * FROM medical_pipeline.gold.kpi_avg_claim_cost_by_payer;
```
