```python
from pyspark.sql.functions import col

dim_procedure = spark.table("medical_pipeline.silver.procedures") \
    .select(
        col("code").alias("procedure_code"),
        col("description")
    ) \
    .dropDuplicates(["procedure_code"])
```

```python
dim_procedure.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("medical_pipeline.gold.dim_procedure")
```
