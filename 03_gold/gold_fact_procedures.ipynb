```python
%sql
create or replace table `medical_pipeline`.gold.fact_procedures as 
with table1 as(
  select * from medical_pipeline.silver.procedures
),table2 as (
  select * from  medical_pipeline.silver.encounters_silver
),table3 as (
  select p.encounter_id ,
  p.patient as patient_id ,
  p.code as procedure_code,
  coalesce(p.base_cost,0) as base_cost,
  to_date(p.start) as procedure_date
  from table1 p left join table2 e on p.encounter_id = e.id
)select * from table3;
```
