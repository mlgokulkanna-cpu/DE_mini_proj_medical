```python
%sql
create or replace table medical_pipeline.gold.dim_dates as
with table1 as(
  select min(start) as min_date,
  max(stop) as max_date
  from medical_pipeline.medical_pipeline.gold.kpi_30day_readmission_ratesilver.encounters_silver
),
table2 as(
  select explode(sequence(min_date , max_date,INTERVAL 1 day)) as date
  from table1
), table3 as(
  select date(`date`) as date,
  year(`date`) as `year`,
  month(`date`) as `month`,
  quarter(`date`) as `quarter`,
  case 
  when quarter(`date`)<=2 then 1
  else 2
  end as half_year
  from table2
)select * from table3 order by `date`;
```
